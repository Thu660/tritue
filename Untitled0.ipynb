```python
#Kiểm tra số lượng ảnh và nhãn trong từng tập dữ liệu: train, valid và test
#sau khi tải về từ Roboflow.
import os

# Đường dẫn gốc sau khi giải nén
base_dir = '/content/GiaoThong-3'  # Đổi path nếu khác

# Kiểm tra các thư mục dữ liệu
train_img_dir = os.path.join(base_dir, 'train/images')
train_lbl_dir = os.path.join(base_dir, 'train/labels')
val_img_dir = os.path.join(base_dir, 'valid/images')  # roboflow dùng 'valid', không phải 'val'
val_lbl_dir = os.path.join(base_dir, 'valid/labels')
test_img_dir = os.path.join(base_dir, 'test/images')
test_lbl_dir = os.path.join(base_dir, 'test/labels')

# Hàm kiểm tra số file trong thư mục
def check_folder(name, img_path, lbl_path):
    img_files = os.listdir(img_path)
    lbl_files = os.listdir(lbl_path)
    print(f'{name}: {len(img_files)} ảnh, {len(lbl_files)} nhãn')

# Kiểm tra dữ liệu
check_folder('Train', train_img_dir, train_lbl_dir)
check_folder('Validation', val_img_dir, val_lbl_dir)
check_folder('Test', test_img_dir, test_lbl_dir)
```

```python
#Cập nhật lại đường dẫn trong file data.yaml cho đúng với cấu trúc thư mục thực
# tế sau khi tải dữ liệu từ Roboflow về.
import os
import yaml

# Đường dẫn thư mục đã giải nén
extract_path = "/content/GiaoThong-3"
yaml_path = os.path.join(extract_path, "data.yaml")

# Mở và chỉnh sửa nội dung
with open(yaml_path, 'r') as f:
    data = yaml.safe_load(f)

# Cập nhật đường dẫn đúng với cấu trúc thư mục trong Colab
data['train'] = "train/images"

# Nếu tồn tại thư mục valid và test thì giữ lại, không thì xóa
valid_path = os.path.join(extract_path, "valid")
test_path = os.path.join(extract_path, "test")

if os.path.exists(valid_path):
    data['val'] = "valid/images"
else:
    data.pop('val', None)

if os.path.exists(test_path):
    data['test'] = "test/images"
else:
    data.pop('test', None)

# Ghi đè lại file
with open(yaml_path, 'w') as f:
    yaml.dump(data, f)

print("Đã sửa xong data.yaml")
```

```python
from ultralytics import YOLO
from PIL import Image
import cv2
import requests
import numpy as np
from io import BytesIO
from IPython.display import display

# Đường dẫn đến file weights tốt nhất của bạn
weights_path = '/content/runs/detect/train/weights/best.pt'

# Đường dẫn ảnh (có thể là URL hoặc đường dẫn cục bộ)
image_path = "https://special.vietnamplus.vn/wp-content/uploads/2021/03/ttxvnhanoi-1587478150-42.jpg"

# Tải mô hình YOLO
model = YOLO(weights_path)

# Đọc ảnh
if image_path.startswith('http'):
    response = requests.get(image_path)
    response.raise_for_status()
    image = Image.open(BytesIO(response.content)).convert("RGB")
else:
    image = Image.open(image_path).convert("RGB")

# Thực hiện nhận diện đối tượng trên ảnh
results = model(image)

# Hiển thị kết quả (ảnh đã được vẽ bounding box) trong Colab
for r in results:
    im_array = r.plot()  # Trả về ảnh dưới dạng mảng numpy BGR
    im_rgb = cv2.cvtColor(im_array, cv2.COLOR_BGR2RGB) # Chuyển BGR sang RGB cho OpenCV
    pil_im = Image.fromarray(im_rgb) # Chuyển từ OpenCV sang PIL
    display(pil_im)

# Lưu ảnh kết quả (tùy chọn)
    pil_im.save('detected_image.jpg')
```

```python
from ultralytics import YOLO
from PIL import Image
import cv2
import requests
import numpy as np
from io import BytesIO
from IPython.display import display, Video
import os
import urllib3

urllib3.disable_warnings(urllib3.exceptions.InsecureRequestWarning)

weights_path = '/content/runs/detect/train/weights/best.pt'
model = YOLO(weights_path)
input_path = "/content/8358430601016742568.mp4"

def detect_video(video_path):
    print("🎥 Bắt đầu detect video...")
    if video_path.startswith('http'):
        try:
            video_data = requests.get(video_path, verify=False).content
            with open("temp_video.mp4", "wb") as f:
                f.write(video_data)
            video_path = "temp_video.mp4"
            print("✅ Đã tải video về local 'temp_video.mp4'")
        except Exception as e:
            print("❌ Lỗi khi tải video:", e)
            return

    cap = cv2.VideoCapture(video_path)
    if not cap.isOpened():
        print(f"❌ Không mở được video: {video_path}")
        return

    width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
    height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
    fps = cap.get(cv2.CAP_PROP_FPS)

    if width == 0 or height == 0:
        width, height = 640, 480
    if fps == 0 or fps is None:
        fps = 30

    out = cv2.VideoWriter('output_video.mp4', cv2.VideoWriter_fourcc(*'mp4v'), fps, (width, height))

    frame_count = 0
    detected_frames = 0
    while cap.isOpened():
        ret, frame = cap.read()
        if not ret:
            break

        frame_count += 1
        results = model(frame)
        draw_frame = frame.copy()
        detected = False
        for r in results:
            if len(r.boxes) > 0:
                detected = True
                im_array = r.plot()
                draw_frame = im_array
        out.write(draw_frame)
        if detected:
            detected_frames += 1

    cap.release()
    out.release()
    print(f"✅ Xong detect video, tổng frame: {frame_count}, frame có đối tượng: {detected_frames}")

detect_video(input_path)
```
